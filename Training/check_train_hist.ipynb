```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("training_data.csv")
#histograma of age
plt.hist(df['Age'], bins=30, color='blue', alpha=0.7)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution in Training Data')
plt.grid()
plt.show()
```

```python
#eliminar de forma aleatoria 800 sujetos que tengan entre 18 y 24 años
young_df = df[(df['Age'] >= 18) & (df['Age'] <= 24)]
other_df = df[(df['Age'] < 18) | (df['Age'] > 24)]
young_df = young_df.sample(n=1200, random_state=42)
balanced_df = pd.concat([other_df, young_df])
plt.hist(balanced_df['Age'], bins=30, color='blue', alpha=0.7)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution in Training Data')
plt.grid()
plt.show()
```

```python
balanced_df.to_csv("training_data.csv", index=False)
```
