```python
import pandas as pd
```

```python
all_data = pd.read_csv("all_data.csv")
val_ids = pd.read_csv("val_ids.csv")
train_ids = pd.read_csv("train_ids.csv")
test_ids = pd.read_csv("test_ids.csv")
```

```python
#columna ID es el valor de la columna "Path" despues del cuarto "/" y antes del ".nii.gz"
all_data["ID"] = all_data["Path"].apply(lambda x: x.split("/")[4].split(".")[0])
all_data = all_data[all_data["ID"] != "run-01"] #saco repetidos
all_data = all_data[~all_data["ID"].str.contains("PT")] #elimino fomo
all_data["ID"] = all_data["ID"].str.replace("_T1w", "")
```

```python
training_merged=pd.merge(train_ids, all_data, on="ID", how="left")
#mostrar los que no se pudieron mergear
training_merged[training_merged["Path"].isnull()]
```

```python
training_merged.to_csv("training_nico.csv", index=False)
```

```python
val_merged=pd.merge(val_ids, all_data, on="ID", how="left")
#mostrar los que no se pudieron mergear
val_merged[val_merged["Path"].isnull()]
```

```python
val_merged.to_csv("val_nico.csv", index=False)
```

```python
all_data = all_data[~all_data["ID"].str.contains("_run-01")]
#si hay un ID que posee la palabra _run-03 elimina el resto de los run-0x y deja solo el _run-03
all_data = all_data[~all_data["ID"].str.contains("_run-0[1-2]")]

#en este codigo cambia "_run-02" por "" en los ids
all_data["ID"] = all_data["ID"].str.replace("_run-02", "")
all_data["ID"] = all_data["ID"].str.replace("_run-03", "")
all_data["ID"] = all_data["ID"].str.replace("_ses-", "_sess-")

all_data["ID"] = all_data["ID"].str.replace(
    r"^sub-(OAS\d+)_sess-(d\d+)$",
    r"\1_MR_\2",
    regex=True
)
```

```python
#filas que en ID contienen "OAS"
all_data[all_data["ID"].str.contains("OAS")].to_csv("oasdebug.csv", index=False)
```

```python
test_merged=pd.merge(test_ids, all_data, on="ID", how="left")
#mostrar los que no se pudieron mergear
test_merged[test_merged["Path"].isnull()]
```

```python
test_merged.to_csv("test_nico.csv", index=False)
```
