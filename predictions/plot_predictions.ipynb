```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("test_predictions_5.csv")
```

```python
plt.figure(figsize=(12, 12))
sns.scatterplot(data=df, y='Age', x='Prediction', color='red')

plt.plot([0, 100], [0, 100], color='black')
plt.title('Age vs Prediction', fontsize=22)
plt.xlabel('Predicted Brain Age', fontsize=20)
plt.ylabel('Age', fontsize=20)

plt.grid(True)

plt.xlim(0, 100)
plt.ylim(0, 100)
plt.tick_params(axis='both', labelsize=16)  # Cambia 16 por el tamaño que prefieras

plt.tight_layout()
plt.show()
print(df['Absolute Error'].mean())
```

```python
# -- Error en funcion de Age --
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Age', y='Error', color='blue')
plt.axhline(0, color='black', linestyle='--')
plt.title('Prediction Error vs Age', fontsize=22)
plt.xlabel('Age', fontsize=20)
plt.ylabel('Prediction Error', fontsize=20)
plt.grid(True)
plt.tick_params(axis='both', labelsize=16)  # Cambia 16 por el tamaño que prefieras
plt.show()
```
