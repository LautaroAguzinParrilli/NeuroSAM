```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
#leer todos los archivos tsv de la carpeta tsvs y concatenarlos en un solo dataframe
import os
tsv_folder = "tsvs"
tsv_files = [f for f in os.listdir(tsv_folder) if f.endswith(".tsv")]
dfs = []
for tsv_file in tsv_files:
    df = pd.read_csv(os.path.join(tsv_folder, tsv_file), sep="\t")
    dfs.append(df)
concat_df = pd.concat(dfs, ignore_index=True)
```

```python
concat_df.sort_values("corr_mean", inplace=True)
concat_df.to_csv("concat_qc.tsv", sep="\t", index=False)
```

```python
#histograma de corr_mean
plt.hist(concat_df["corr_mean"], bins=20)
plt.xlabel("corr_mean")
plt.ylabel("Frecuencia")
plt.title("Histograma de corr_mean")
plt.ylim(0, 100)
plt.show()
```
