```python
import pandas as pd
import os
```

```python
df= pd.read_csv("../Data/all_participants.csv")

# Create a DataFrame from the file paths
folder_path = "/data/lautaro/quasiraw_ext/papers" #cambiar a training

data = []
for root, dirs, files in os.walk(folder_path):
    for file in files:
        file_path = os.path.join(root, file)
        data.append({'ID': file.split('.nii.gz')[0], 'Path': file_path})

file_paths = pd.DataFrame(data)
file_paths['ID'] = file_paths['ID'].str.replace('.nii.gz', '')
file_paths['ID'] = file_paths['ID'].str.replace('_T1w', '')
```

```python
print(file_paths)
```

```python
csv_data = pd.merge(df, file_paths, on='ID', how='right')[['Path', 'Age']]
```

```python
#ver cuales son las filas con valores nulos
null_rows = csv_data[csv_data.isnull().any(axis=1)]
print("Filas con valores nulos:")
print(null_rows)
```

```python
csv_data['Age'][csv_data['Age'].isnull()]=0.0
```

```python
#guardar csv
csv_data.to_csv("papers_test_data.csv", index=False)
```
