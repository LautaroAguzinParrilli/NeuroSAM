```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fomo=pd.read_csv('all_fomo_participants.csv')
df=pd.read_csv('all_participants.csv')
```

```python
df_concat=pd.concat([fomo, df]).reset_index(drop=True)
```

```python
#histograma de edades
plt.hist(df_concat['Age'].dropna(), bins=20, edgecolor='black')
plt.title('Age Distribution of All Participants')
plt.xlabel('Age')
plt.ylabel('Number of Participants')
plt.show()
```
