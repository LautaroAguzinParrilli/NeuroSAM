```python
import pandas as pd
```

```python
adnidata=pd.read_csv('ADNIparticipants.csv')
unsamdata=pd.read_csv('UNSAMparticipants.csv')
ext_adni=pd.read_csv('ext_test_ADNI.csv')
ext_unsam=pd.read_csv('ext_test_UNSAMLC.csv')
```

```python
# remplazar /data/lautaro/quasiraw_ext/papers/ de la columna Path
ext_adni['ID'] = ext_adni['Path'].str.replace('/data/lautaro/quasiraw_ext/papers/', '')
ext_adni['ID'] = ext_adni['ID'].str.replace('.nii.gz', '')

ext_unsam['ID'] = ext_unsam['Path'].str.replace('/data/lautaro/quasiraw_ext/papers/', '')
ext_unsam['ID'] = ext_unsam['ID'].str.replace('.nii.gz', '')
```

```python
adnidata=adnidata[['ID', 'Group']]
unsamdata=unsamdata[['ID', 'Grupo']]
```

```python
ext_adni_CNs = ext_adni.merge(adnidata, on='ID', how='inner')
ext_unsam_CNs = ext_unsam.merge(unsamdata, on='ID', how='inner')
```

```python
ext_unsam_CNs=ext_unsam_CNs[ext_unsam_CNs['Grupo']=='CONTROL']
ext_adni_CNs=ext_adni_CNs[ext_adni_CNs['Group']=='CN']

ext_adni_CNs.drop(columns=['ID','Group'], inplace=True)
ext_unsam_CNs.drop(columns=['ID','Grupo'], inplace=True)
```

```python
ext_unsam_CNs.to_csv('ext_test_UNSAMLC_CN.csv', index=False)
ext_adni_CNs.to_csv('ext_test_ADNI_CN.csv', index=False)
```
